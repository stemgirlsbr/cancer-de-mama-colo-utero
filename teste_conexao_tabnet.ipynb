```python
import socket, ftplib, urllib.request

for host in ["ftp.datasus.gov.br", "tabnet.datasus.gov.br"]:
    try: print("DNS ok:", host, socket.gethostbyname(host))
    except Exception as e: print("DNS FALHOU:", host, e)

try:
    ftp = ftplib.FTP("ftp.datasus.gov.br", timeout=30); ftp.login()   # testes 2 e 3
    print("Porta 21 ok")
    ftp.cwd("/dissemin/publicos"); print("Listagem ok:", ftp.nlst()[:5])  # usa modo passivo
    ftp.quit()
except Exception as e: print("FTP FALHOU:", type(e).__name__, e)

for url in ["http://tabnet.datasus.gov.br/", "https://tabnet.datasus.gov.br/"]:  # teste 4
    try: print(url, urllib.request.urlopen(url, timeout=30).status)
    except Exception as e: print(url, "FALHOU:", type(e).__name__, e)
```
